# Objective 

To create a comprehensive and well-documented tutorial that introduces the computational fluid dynamics (CFD) simulation tools used at the STFS institute. This tutorial aims to facilitate the onboarding process for future students and employees, enabling them to efficiently conduct research in the field of energy conversion systems. The tutorial will cover a wide range of methods and tools, including 1D/2D non-premixed flames, triple flame configurations, and post-processing routines in Python. The ultimate goal is to provide a valuable resource that aligns with STFS's research data management guidelines, presenting and discussing all results.


# Description 

The "Development of a Tutorial for the Simulation of Future Sustainable Propulsion Systems" project, conducted at the Institute of Simulation of Reactive Thermal Fluid Systems (STFS), addresses the increasing significance of Computational Fluid Dynamics (CFD) simulations in the design of future energy conversion systems.  
The tutorial covers a wide array of pertinent methods and tools, including the calculation of 1D/2D non-premixed flames (NP) using Cantera, the in-house solver ULF, and OpenFOAM. It also encompasses the setup of the triple flame configuration. Additionally, the tutorial provides guidance on post-processing routines in Python. All simulation results, particularly in the context of reactive thermal fluid systems, are presented and discussed comprehensively.  
This tutorial project represents a substantial contribution to the field of reactive thermal fluid systems, empowering future researchers to harness advanced CFD simulations effectively. It serves as an invaluable resource for the STFS community and the broader scientific community, supporting advancements in the design and development of sustainable and efficient thermal fluid systems.
