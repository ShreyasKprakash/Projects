# Prerequisites

### Theory

Thermodynamics: Understand the principles of thermodynamics, including concepts like energy conservation, the First and Second Laws of Thermodynamics, thermodynamic cycles, and properties of gases and fluids. These are crucial for analyzing energy conversion systems.

Heat Transfer: Knowledge of heat transfer mechanisms, such as conduction, convection, and radiation, is important for understanding the thermal aspects of combustion and energy conversion.

Fluid Dynamics: Familiarity with fluid dynamics, including concepts like fluid flow, pressure, velocity, and conservation equations (e.g., Navier-Stokes equations), is essential for CFD simulations.

### Basics
##### Python: 
Python is often used for post-processing and scripting in CFD simulations. Should have a basic understanding of Python for data analysis and scripting tasks.

##### OpenFOAM: 
OpenFOAM is an open-source CFD software package. Should have a good grasp of how to use OpenFOAM for setting up and running simulations.

 Step 1: Install Windows Subsystem for Linux [here](https://learn.microsoft.com/en-us/windows/wsl/install)    
* Install WSL by following the instructions in the "Install WSL" section.       Make sure to choose a Linux distribution (e.g., Ubuntu) during the installation process.  
 * After the installation is complete, there should be Ubuntu (or your chosen Linux distribution) as a Windows program installed.

 * To install go to the official Visual Studio Code website:[here](https://code.visualstudio.com/)
 
 Step 3: Install the required extensions in Visual Studio Code.

* Connect to VPN to access the cluster:

  * Once connected to the login node, can perform tasks like submitting batch jobs to the cluster's compute nodes, accessing data, and running interactive applications or code within the cluster's environment. The login nodes act as the gateway to the cluster's resources and provide a secure way for users to interact with the cluster

# Paraview


paraview is a Post Processing tool and is an open-source, multi-platform data analysis and visualization application. It is primarily used for visualizing and analyzing scientific and engineering data from a wide range of sources, including computational simulations, experimental measurements, and observational data.

steps to use paraview.
* Start tigervnc on local machine (Windows, not WSL in code)  
* Right click on desktop in tigervnc and start terminal  
* Load paraview module  
* Run from terminal `vglrun paraview` inside your OF case → you need to have a .foam file in this directory, just generate one with `touch example_file.foam`
* Open file in paraview and select the .foam file
