### 1.1. Motivation<a class="anchor" id="motivation"></a>

Within the framework of Industry 4.0, manufacturing machines are increasingly being equipped with sensors that can automatically generate and store data in the production processes. <br>
Utilizing data from sensors in the CNC milling machine, the aim is to gain valuable insights into the manufacturing process. The driving motivation behind detecting anomalies in the data is to ensure product quality and proactively address potential issues during the production of cylinder bottoms. This approach not only enhances production efficiency but also provides a foundation for leveraging machine learning (ML) algorithms. ML allows for predictive analytics, enabling the identification of irregularities before they impact the final product, leading to improved reliability and cost-effectiveness in the manufacturing process.

### 1.2. Objective<a class="anchor" id="objective"></a>

Detect anomalies of the cylinder-bottom parts using processing data from the CNC milling machine. The output of your analysis shall be a binary classification of the cylinder bottom parts:  <br> 
False: no anomaly (anomaly label 0)<br>
True: detected anomaly (anomaly labels 1, 2 and 3)<br>
The f1-score of the True class will be used for evaluation of results.

### 1.3. Approach<a class="anchor" id="approach"></a>

**CRISP-DM (Cross Industry Standard Process model for Data Mining)**

<div style="text-align:center;">
    <img src="MLA_Figures/1.jpg" alt="2" style="width:50%; max-width:800px;">
</div>
<center>Figure 1: CRISP-DM Model</center> 

In order to ensure that, CRISP-DM approach to the data challenge is thorough and well-organized, decided to utilize its framework. This framework is based on the idea that data mining projects can be made more efficient, reliable, repeatable, manageable, and faster by following a comprehensive and adaptable process model. The CRISP-DM framework features a hierarchical process model that comprises four levels of abstraction, including phases, generic tasks, specialized tasks, and process instances.At the heart of the CRISP-DM framework are six phases that are designed to help to achieve the objectives in a structured and effective way. 

1. `Business Understanding` : it involves defining the goals and objectives. This is a critical step in ensuring that the given task is aligned with the strategic objectives and a clear understanding of what is to be achieved.<br>

2. `Data Understanding`: which involves analyzing the available/given data and assessing its quality to determine its suitability for analysis. This phase helps to gain a better understanding of the type of data working with, as well as any limitations or challenges that may need to be addressed.<br>

3. `Data Preparation` : this phase which filtering and selecting relevant data and preparing it for modeling. This phase is critical in ensuring that the data is in a format that can be used for analysis and that it is of sufficient quality to produce reliable results.<br>

4. `Modeling`: in this stage models are created to meet the  defined task goals. This phase involves selecting the appropriate modeling techniques and algorithms to apply to the data and fine-tuning the models to achieve the desired outcomes.<br>

5. `Evaluation`: the models' performance is assessed to determine their effectiveness and accuracy. This phase helps the team understand how well the models are performing and identify any areas where improvements can be made.<br>

6. `Deployment`: involves integrating the model into existing systems and monitoring its performance over time. This phase is critical in ensuring that the model continues to perform well and that any issues that arise are quickly addressed.<br>

Overall, the CRISP-DM framework provides a valuable and effective approach to data mining tasks. By following this framework, one can ensure that the approach is structured, thorough, and effective, leading to the best possible outcome and improved overall performance.

Findings relevant to the development of the machine learning model include insights drawn from the visualization of timestamps. 2. Random Forest: Random Forest is chosen due to the feature importance measures, which help to interpret the realtive importance of different features in making predictions.

Statistical features such as mean, root mean square (RMS), standard deviation and variance were usually used in past studies[1-2] to identify the differences between one vibration signal and another. These features are called time-domain dimensional features, because they are calculated directly from the vibration signals (in timedomain).


 More advanced statistical-based features such as skewness and kurtosis can be applied to the signal which is not purely stationary. These features examine the probability density function (PDF) of the signal. It is a well-known fact that if the condition of the tool changes, the PDF also changes, thus the skewness and kurtosis might be able to detect the changes. Furthermore, skewness is used to measure whether the signal is negatively or positively skewed. Kurtosis measures the peak value of the PDF and indicates that the signal is impulsive in nature. Based on the definition of moment, skewness is equal to zero for a signal with a normal distribution i.e. normal bearing signal. A negative value is due to skewness towards lower value while a positive value indicates non-symmetry towards higher values.

 when the vibration signal changes due to faults, the kurtosis value will increase to greater than three and the skewness value will shift to either negative or positive. In higher order statistics (HOS), the kurtosis and skewness based on the definition of cumulant have value zero for Gaussian signals. Other features which can be calculated from PDF of the vibration signal includes: entropy, which calculates the histogram of the PDF and measures the degree of randomness of the vibration signal; lower bound and upper bound histogram, which measure the lower and upper values of the PDF respectively. Aside from the time-domain dimensional features, there are other non-dimensional features such as shape factor (defined as RMS divided by mean) and crest factor.(defined as standard deviation divided by RMS).
